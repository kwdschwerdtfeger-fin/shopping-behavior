```python
# Previous Purchases is unclear. We don't know if it was for the same user.
# We can work around this by aggregating information from multiple columns
# We don't know if the gender and age refer to individuals or groups
# For now, we will make a pivot table

shopping_data.pivot_table(values = 'Purchase Amount (USD)', index = 'Color', columns = 'Season', aggfunc = 'sum').reset_index().sort_values('Summer', ascending = False)

#This table shows the data is spread across four columns corresponding to the seasons of one year (Fall, Spring, Summer, Winter),
#and each season is treated as a singular categorical variable, not a sciding scale (timewise).

#The table also shows all colors present in the data, and therefore utilizd by the company in question.
```

```python
'''The code below ranks the popularity of clothing items by color during the Winter season.'''
shopping_data.pivot_table(values = 'Purchase Amount (USD)', index = 'Color', columns = 'Season', aggfunc = 'sum').reset_index().sort_values('Winter', ascending = False)
```

**Conclusions from code cells above: Purchase by season**
- For the Fall season, ***brown*** is the most commonly purchased color.
- For the Spring season, ***baby blue*** is the most commonly purchased color.
- For the Summer season, ***lavender*** is the most commonly purchased color.
- For the Winter season, ***black*** is the most commonly purchased color.

***PROMO CODE EFFECT***

The code below is clarifying the effect of promo codes on the dollar amount of purchases.

```python
shopping_data.groupby('Promo Code Used')['Purchase Amount (USD)'].mean()
# shopping_data.pivot_table(values = 'Purchase Amount (USD)', index = 'Item Purchased', columns = 'Promo Code', aggfunc = 'sum').reset_index().sort_values('No', ascending = False)
```

***PROMO CODE CONCLUSION***

As we can tell, there is a notable difference in the purchase amount of all goods when promo codes are used, specifically regarding the difference in means.
